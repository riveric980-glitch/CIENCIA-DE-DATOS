```python
def clasificar_edad(edad):
    """recibe una edad y regresa la categoria correspondiente."""
    if edad < 12:
        return "niño"
    elif edad < 18 :
        return"adolecente"
    elif edad < 65:
        return "adulto"
    else: 
        return "adulto mayor"
        #probar funcion con distintis valores
for edad in [5, 15, 40, 80]:
    print (edad, "->", clasificar_edad(edad))
```
