```python
#imptotatr la librerias que vamos a ocuapr
import pandas as pd #carga datos
import numpy as np # analisis  de dtsoa
import matplotlib.pyplot as plt #graficos
```

```python
#cargar archivo de datos#data framw"
df = pd.read_csv('titanic.csv')
print("carga dedatos exitosa!!")
```

limpieza basica de datos
tratar valores faltantes
droppear
convertir datos

```python
df['Age'] = df['Age'].fillna(df['Age'].median()) #Rellenar columnas de Age
```

```python
df.head()
```

```python
df.dropna(subset=['Embarked'], inplace=True)
```

```python
df.drop('Cabin', axis=1, inplace=True)
```

```python
df['Pclass'] = df['Pclass'].astype('category') #convertir la columna
```

```python
df.info()
```

```python
import seaborn as sns
sns.set(style="whitegrid" , palette="muted")
plt.rcParams["figure.figsize"]= (10, 6)
```

```python
sns.countplot( x="Survived" , data=df)
plt.title("Distribucion de sobrevivientes")
plt.show()
```

```python
cat_vars = ["Sex", "Pclass", "Embarked"]
```

```python
for col in cat_vars:
    sns.countplot(x=col, data=df)
    plt.title("Distribucion de {col}")
    plt.show()
```

```python
#variables numericas
num_vars= ["Age", "Fare"]
```

```python
for col in num_vars:
    sns.histplot(df[col].dropna(), kde=True, bins=30)
    plt.title("Distribucion de {col}")
    plt.show()
```

analisis dos variable

```python
#supervivencia por sexo
sns.barplot(x="Sex", y="Survived", data=df)
plt.title("Sobrevivientes segun sexo")
plt.show()
```

```python
#supervivencia por clase
sns.barplot(x="Pclass", y="Survived", data=df)
plt.title("Sobrevivientes egun clse")
plt.show()
```

```python
#suspervivencia por edad
sns.barplot(x="Survived", y="Age", data=df)
plt.title("Sobrevivientes segun edad")
plt.show()
```

```python
#correlaciones
corr = df.corr(numeric_only=True)
sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Mapa de corelaciones")
plt.show()
```
